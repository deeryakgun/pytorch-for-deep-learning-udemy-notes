## Reshaping, stacking, squeezing, unsqueezing tensors

```python
x = torch.arange(1,10)
x
```

```python
#add an extra dimension
x_reshaped = x.reshape(1,9)
x_reshaped , x
```

```python
z = x.view(1,9) #z is different view of x
z
```

```python
#changing z change x because a view of a tensor shares the same memory as the original input.
z[0,0] = 5
z,x
```

```python
x_stacked = torch.stack([x,x,x,x], dim=0) #stack vertical
x_stacked
```

```python
torch.stack([x,x,x], dim=1) #stack horizantal
```

```python
# tensor.squeeze() - removes all single dimensions from a target tensor
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

x_sq = x_reshaped.squeeze()
print(f"\nNew tensor: {x_sq}")
print(f"New shape: {x_sq.shape}")
```

```python
#tensor.unsqueeze() - adds a single dimension to target tensor at a specific dim ()
print(f"Previous tensor: {x_sq}")
print(f"Previous shape: {x_sq.shape}")

x_unsq = x_sq.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsq}")
print(f"New shape: {x_unsq.shape}")
```

```python
#tensor.permute() - rearrenges the dimensions of a target tensor to a specified order
x_original = torch.rand(size=[224,224,3]) #height, width, color channels

x_permuted = x_original.permute(2,0,1)   #color channels, height, width

print(f"Original tensor shape: {x_original.shape}")
print(f"Permuted tensor shape: {x_permuted.shape} ")
```

```python
x_original[0,0,0] = 3
x_original[0,0,0], x_permuted[0,0,0] #share same memory
```
